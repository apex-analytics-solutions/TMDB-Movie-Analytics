## Creating JSON Extraction Function

A reusable function is created to extract meaningful names from JSON formatted columns.

```python
def extract_names(json_string):
    """
    Extract names from JSON formatted strings
    """

    try:
        data = ast.literal_eval(json_string)

        names = [
            item["name"]
            for item in data
        ]

        return ", ".join(names)

    except:
        return None
```

```python
# Test extraction function

extract_names(
    movies_clean.loc[0,"genres"]
)
```

# 5. Feature Engineering

Feature engineering transforms raw data into meaningful analytical variables.

In this section, new features are created to support deeper business analysis:

- Time-based features from release dates
- Financial performance metrics
- Investment return indicators

```python
# Convert release_date into datetime format

movies_clean["release_date"] = pd.to_datetime(
    movies_clean["release_date"],
    errors="coerce"
)


movies_clean["release_date"].head()
```

## Creating Time-Based Features

Release dates are transformed into year, month, and decade features to analyze historical movie trends.

```python
# Extract release year

movies_clean["release_year"] = (
    movies_clean["release_date"]
    .dt.year
)


movies_clean[
    [
        "title",
        "release_date",
        "release_year"
    ]
].head()
```

```python
# Extract release month

movies_clean["release_month"] = (
    movies_clean["release_date"]
    .dt.month
)


movies_clean[
    [
        "title",
        "release_month"
    ]
].head()
```

```python
# Create decade feature

movies_clean["release_decade"] = (
    (movies_clean["release_year"] // 10) * 10
).astype("Int64")


movies_clean["release_decade"] = (
    movies_clean["release_decade"]
    .astype(str)
    + "s"
)


movies_clean[
    [
        "title",
        "release_year",
        "release_decade"
    ]
].head()
```

## Financial Feature Creation

Financial metrics help measure movie performance beyond simple revenue.

New metrics:

- Profit = Revenue - Budget
- ROI = Profit / Budget

These features help evaluate profitability and investment efficiency.

```python
# Calculate movie profit

movies_clean["profit"] = (
    movies_clean["revenue"]
    -
    movies_clean["budget"]
)


movies_clean[
    [
        "title",
        "budget",
        "revenue",
        "profit"
    ]
].head()
```

```python
# Calculate ROI

movies_clean["ROI"] = np.where(
    movies_clean["budget"] > 0,
    movies_clean["profit"] / movies_clean["budget"],
    np.nan
)


movies_clean[
    [
        "title",
        "budget",
        "revenue",
        "profit",
        "ROI"
    ]
].head()
```

## Financial Dataset Preparation

Movies without budget or revenue information cannot provide reliable financial insights.

Therefore, only movies with valid financial records are selected.

```python
# Filter valid financial records

movies_financial = movies_clean[
    (movies_clean["budget"] > 0) &
    (movies_clean["revenue"] > 0)
]


print("Original Dataset:", movies_clean.shape)

print("Financial Dataset:", movies_financial.shape)
```

```python
movies_clean[
    [
        "title",
        "release_year",
        "release_decade",
        "profit",
        "ROI"
    ]
].head()
```
